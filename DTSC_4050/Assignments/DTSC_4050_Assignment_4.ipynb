## 30 Points

The dataset below represents customers applying for a service. The target `renewed` indicates whether the customer renewed:

- `1` = Renewed
- `0` = Did not renew

You will build and evaluate a **k-nearest neighbors (k-NN)** classifier. This section assesses concepts practiced in the classification notebook: features/classes, standardization, training/test sets, choosing `k`, accuracy, and classification errors.


```python
rng = np.random.default_rng(4050)
n = 160

customer_data = pd.DataFrame({
    "monthly_usage": rng.normal(45, 15, n).clip(5, 90),
    "support_calls": rng.poisson(3, n).clip(0, 10),
    "months_as_customer": rng.integers(1, 61, n)
})

renewal_score = (
    0.055 * customer_data["monthly_usage"]
    - 0.40 * customer_data["support_calls"]
    + 0.035 * customer_data["months_as_customer"]
    - 2.0
    + rng.normal(0, 0.8, n)
)

customer_data["renewed"] = (renewal_score > 0).astype(int)
customer_data.head()
```
